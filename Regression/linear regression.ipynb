```python
X=df['YearsExperience']
Y=df['Salary']
```

```python
# Import scikit learn learn library
from sklearn.model_selection import train_test_split
```

```python
# Split the train and test dataset
X_train,X_test,y_train,y_test = train_test_split(X,Y, train_size=0.7,random_state=42)
```

```python
# Convert Pandas Series → 2D array
X_train = X_train.values.reshape(-1, 1)
X_test  = X_test.values.reshape(-1, 1)
```

```python
#Importing the model
from sklearn.linear_model import LinearRegression
```

```python
#Fitting data to the model
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
#Predicting the Salary for the test values
y_predict = model.predict(X_test)
y_predict
```

```python
model.predict([[3.4]])
```

```python
#Plotting the actual and predicted values
c = [i for i in range (1,len(y_test)+1,1)]
plt.plot(c,y_test,color='r',linestyle='-')     # is this line is a predict salary value
plt.plot(c,y_predict,color='b',linestyle='-')  # is this line is a actual salary value
plt.xlabel('index')
plt.ylabel('salary')
plt.title('Actual vs. Predicted')
```

```python

#Plotting the error
c = [i for i in range(1,len(y_test)+1,1)]
plt.plot(c,y_test-y_predict,color='green',linestyle='-')    #Error = Actual value − Predicted value
plt.xlabel('index')
plt.ylabel('Error')
plt.title('Error Value')
```

```python
#Intercept and coeff of the line
print('Intercept of the model:',model.intercept_)
print('Coefficient of the line:',model.coef_)
```
